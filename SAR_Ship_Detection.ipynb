```python
import os
import json
import shutil
from tqdm import tqdm

DATASET_PATH = "/root/.cache/kagglehub/datasets/kailaspsudheer/sarscope-unveiling-the-maritime-landscape/versions/2/SARscope"

YOLO_PATH = "/content/SARscope_YOLO"

for split in ["train", "valid", "test"]:
    os.makedirs(os.path.join(YOLO_PATH, "images", split), exist_ok=True)
    os.makedirs(os.path.join(YOLO_PATH, "labels", split), exist_ok=True)


def convert_coco_to_yolo(split):

    split_path = os.path.join(DATASET_PATH, split)
    json_path = os.path.join(split_path, "_annotations.coco.json")

    with open(json_path, "r") as f:
        coco = json.load(f)

    images = {
        img["id"]: img
        for img in coco["images"]
    }

    annotations = {}

    for ann in coco["annotations"]:
        image_id = ann["image_id"]

        if image_id not in annotations:
            annotations[image_id] = []

        annotations[image_id].append(ann)

    print(f"\n📂 {split.upper()}")
    print("Images:", len(images))
    print("Annotations:", len(coco["annotations"]))

    for image_id, img_info in tqdm(images.items()):

        filename = img_info["file_name"]

        src_image = os.path.join(split_path, filename)
        dst_image = os.path.join(
            YOLO_PATH,
            "images",
            split,
            filename
        )

        shutil.copy2(src_image, dst_image)

        width = img_info["width"]
        height = img_info["height"]

        label_filename = os.path.splitext(filename)[0] + ".txt"

        label_path = os.path.join(
            YOLO_PATH,
            "labels",
            split,
            label_filename
        )

        lines = []

        for ann in annotations.get(image_id, []):

            x, y, w, h = ann["bbox"]

            x_center = x + w / 2
            y_center = y + h / 2

            x_center /= width
            y_center /= height

            w /= width
            h /= height


            class_id = 0

            lines.append(
                f"{class_id} "
                f"{x_center:.6f} "
                f"{y_center:.6f} "
                f"{w:.6f} "
                f"{h:.6f}"
            )

        with open(label_path, "w") as f:
            f.write("\n".join(lines))


for split in ["train", "valid", "test"]:
    convert_coco_to_yolo(split)

```

```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

MODEL_PATH = "/content/runs/SARscope_YOLO11s/weights/best.pt"

model = YOLO(MODEL_PATH)

image_path = "/content/SARscope_YOLO/images/test"

results = model.predict(
    source=image_path,
    imgsz=640,
    conf=0.25,
    device=0,
    save=False
)

for result in results[:5]:

    annotated = result.plot()

    ship_count = len(result.boxes)

    print(f"Image: {result.path}")
    print(f"🚢 Ships detected: {ship_count}")

    plt.figure(figsize=(10, 8))
    plt.imshow(cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB))
    plt.axis("off")
    plt.show()
```

```python
from collections import Counter

test_results = model.predict(
    source="/content/SARscope_YOLO/images/test",
    imgsz=640,
    conf=0.25,
    device=0,
    verbose=False
)

ship_counts = []

for result in test_results:
    count = len(result.boxes)
    ship_counts.append(count)

print("Number of test images:", len(ship_counts))
print("Total detected ships:", sum(ship_counts))
print("Average ships/image:", sum(ship_counts) / len(ship_counts))
print("Maximum ships in one image:", max(ship_counts))
print("Minimum ships in one image:", min(ship_counts))
```

```python
all_confidences = []

for result in test_results:

    if result.boxes is not None and len(result.boxes) > 0:

        confidences = result.boxes.conf.cpu().numpy()
        all_confidences.extend(confidences)

if all_confidences:

    print("Average Confidence:",
          sum(all_confidences) / len(all_confidences))

    print("Highest Confidence:",
          max(all_confidences))

    print("Lowest Confidence:",
          min(all_confidences))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.hist(
    ship_counts,
    bins=20
)

plt.xlabel("Number of detected ships")
plt.ylabel("Number of images")
plt.title("Ship Detection Distribution")

plt.show()
```
